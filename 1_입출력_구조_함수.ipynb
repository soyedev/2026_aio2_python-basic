```python
snack_box = ['베스킨']

new_snack = input('먹고 싶은 간식을 입력하세요(, 로 여러 개를 연결하세요): ').split(',')
snack_box += new_snack

# 삭제 전 간식 개수 저장
snack_count = len(snack_box)

# 번호와 함께 목록 출력
print("\n현재 간식 목록")
for i, snack in enumerate(snack_box, start=1):
    print(f"{i}. {snack}")

# 빼고 싶은 간식 번호 입력
remove_num = int(input('빼고 싶은 간식이 있으면 번호를 입력하세요: '))
snack_box.pop(remove_num - 1)

# 주문 수량 입력
box_count = int(input('몇 상자를 주문하시겠어요? (숫자로): '))

# 삭제 전 간식 기준으로 계산
total = snack_count * box_count

print("\n최종 간식 목록:", snack_box)
print("주문 상자 수:", box_count)
print("총 주문 개수:", total)
print("\n주문해주셔서 감사합니다.")
```
