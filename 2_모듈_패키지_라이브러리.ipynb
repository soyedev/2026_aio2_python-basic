```python
# 라이브러리 호출
import pandas as pd

# 표로 보기위한 원시 데이터(딕셔너리)
data = {
    '이름': ['김철수', '이영희', '박민수', '정수진', '최영호', '한미래', '윤지원'],
    '나이': [25, 28, 22, 31, 27, 24, 29],
    '점수': [85, 92, 78, 95, 88, 73, 91]
}
```

```python
# 판다스의 데이터프레임객체 생성
df = pd.DataFrame(data)
# 상위 5개 데이터 조회
#df.head()
```
