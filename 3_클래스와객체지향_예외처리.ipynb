```python
# 클래스 정의
class Calculator:
    def __init__(self, number_list):
        self.number_list = number_list

    def sum(self):
        return sum(self.number_list)

    def avg(self):
        return sum(self.number_list) / len(self.number_list)
```

##  Account 클래스
은행에 가서 계좌를 개설하면 은행이름, 예금주, 계좌번호, 잔액이 설정됩니다.
Account 클래스를 생성한 후 생성자를 구현해보세요.
생성자에서는 예금주와 초기 잔액만 입력 받습니다.
은행이름은 SC은행으로 계좌번호는 3자리-2자리-6자리 형태로 랜덤하게 생성됩니다.

은행이름: SC은행
계좌번호: 111-11-111111

```python
import random

class Account:
    def __init__(self, name, balance):
        self.bank = "SC은행"
        self.name = name
        self.balance = balance
        self.account_number = (
            f"{random.randint(100, 999)}-"
            f"{random.randint(10, 99)}-"
            f"{random.randint(100000, 999999)}"
        )
```

```python
import random


class Account:
    bank_name = 'SC은행' #클래스 변수
   
    def __init__(self, name, balance):
        self.name = name #인스턴스 변수
        self.balance = balance
        self.account_no = Account.make_account_no()

    @staticmethod   
    def make_account_no():  
        num1 = random.randint(0, 999)
        num2 = random.randint(0, 99)
        num3 = random.randint(0, 999999)


        return f"{num1:03d}-{num2:02d}-{num3:06d}"
   
    def print_info(self,):
        return f'은행이름 : {Account.bank_name} \n계좌번호: {self.account_no}'
   
    def 입금(self, amount):
        self.balance += amount
   
    def 출금(self, amount):
        self.balance -= amount
   
    def 잔고조회(self):
        return f' {self.name}님의 잔고는 {self.balance} 입니다'

```

```python
ac1 = Account('홍길동', 2000)
print(ac1.print_info())
ac1.잔고조회()
```

## 상속과 재정의(overriding)

```python
class Animal:
    def make_sound(self):
        pass


class Dog(Animal):
    def make_sound(self): # 재정의 overriding
        return "멍멍!"


class Cat(Animal):
    def make_sound(self): # 재정의 overriding
        return "야옹!"
```

```python
Cat().make_sound()
```

## 다형성

```python
animal = [Dog(), Cat()]

for ani in animal:
    print(type(ani))
    print(ani.make_sound())
```

```python
class AgeException(Exception):
    pass


def input_age():
    age = int(input('나이를 입력하세요: '))

    if age <= 0:
        raise AgeException('나이는 양수여야 합니다.')

    elif age > 20:
        raise AgeException('나이는 20세 이하여야 합니다.')

    return age


# 예외처리를 하는 나이 입력 구현
try :
    age = input_age()
except AgeException as e:
    print(e.args[0])
else :
    print(f'입력하신 나이는 {age}입니다.')
```
